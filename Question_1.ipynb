```python
# information about each csv file seperatly.
customers = pd.read_csv('customers.csv')
Items = pd.read_csv('Items.csv')
order_item = pd.read_csv('order_item.csv')
orders = pd.read_csv('orders.csv')
display('customers', customers,'Items' , Items, 'order_item', order_item, 'orders', orders)
%config Completer.use_jedi = False
```

```python
# Qestion 1:
# part 2- Duplication Deletion:

######## Creating a new DF to replace 'Items', keeping the original as an archive.
Items_no_doubles = Items

######## removing Duplicate's based on greater amount of Non-null values:
Items_no_doubles = (Items.assign(Non_null_count = Items.count(axis=1))
    .sort_values( by = ['item_name', 'Non_null_count'], ascending= [False, False] )
    .drop_duplicates( subset = ['item_name'], keep ='first' ).sort_index())

######## maping id from items, maping replacment id's andreplacing it in order_item:
keep_Item_id_map = Items_no_doubles.set_index('item_name')['id']
Items_maped = Items[['id', 'item_name']].copy()
Items_maped['new_id'] = Items_maped['item_name'].map(keep_Item_id_map)

dict_replace = Items_maped.set_index('id')['new_id'].to_dict()
order_item['item_id'] = order_item['item_id'].map(dict_replace)
display("displaying only corrected item_id's, all Duplication's who has las non-null's had been removed", order_item)

###### if you want to precent ID sorted - apply this code:
    # 1.  check_replacment = order_item['item_id'].reset_index()
    # 2.  check_replacment.groupby('item_id').count() ''')
```

```python
######## Summing the missing values (NaNs) across rows + Creating a new DF to replace 'customers'
cleand_customers = customers[customers.isna().sum(axis=1) <3].copy()
display('all valid ONLY, customers ONLY (DataFrame): ', cleand_customers)

######## filtering all not valid orders:
valid_customers = cleand_customers['id']

valid_Cids_orders = orders[orders['customer_id'].isin(valid_customers)]
valid_orders_ids = valid_Cids_orders['order_id']

######## filtering all not valid order_items:
cleaned_order_item = order_item[order_item['order_id'].isin(valid_orders_ids)]
display("all valid ONLY, order_item ID's (DataFrame): ", cleaned_order_item)
```
